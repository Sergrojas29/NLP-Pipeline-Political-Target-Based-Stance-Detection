```python
import spacy
nlp = spacy.load("en_core_web_trf")
nlp.add_pipe("entityLinker", last=True)
doc = nlp(text)
```

```python
person_spans_text = [(ent.start_char, ent.end_char, ent.text) for ent in doc.ents if ent.label_ == "PERSON" or ent.label_ == "NORP" or ent.label_ ==  "GPE" ]
person_spans_text
```

```python
# ner_spans = [(ent.start_char, ent.end_char) for ent in doc.ents if ent.label_ == "PERSON" or ent.label_ == "NORP" or ent.label_ ==  "GPE"]
person_spans = [(ent.start_char, ent.end_char) for ent in doc.ents if ent.label_ == "PERSON"]
data = {}
for linked_ent in doc._.linkedEntities:
    span = linked_ent.get_span()
  
    is_person = any(start <= span.start_char and span.end_char <= end for start, end in person_spans)
    
    
    if is_person and linked_ent.description != "family name":
        
        id = linked_ent.get_id()
        if id not in data:
            data[linked_ent.get_id()] ={
                "name": linked_ent.get_label(),
                "text": set(),
                "description": linked_ent.description
            }
        data[id]["text"].add(span.text)      
       
for k , (name, text, desx) in data.items():
    print(data[k][name])
data
```
